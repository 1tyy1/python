```python
#1 say_hi：无参，打印Hello, world!
def say_hi():
    print("Hello, world!")

#2 goat_pad：字符串前后拼接goat
def goat_pad(s):
    print("goat" + s + "goat")

#3 print_n：字符串s打印n次，每行一个
def print_n(s, n):
    for _ in range(n):
        print(s)

#测试
print("===测试say_hi===")
say_hi()

print("\n===测试goat_pad===")
goat_pad("bird")
goat_pad("_")

print("\n===测试print_n===")
print_n("Hi",3)
```

```python
def gcd(m,n):
    """欧几里得辗转相除法求最大公约数"""
    while n != 0:
        m, n = n, m % n
    return m

#测试三组数字
print(f"gcd(2018,2019) = {gcd(2018,2019)}")
print(f"gcd(1200,300) = {gcd(1200,300)}")
print(f"gcd(5040,60) = {gcd(5040,60)}")

#测试负数
print(f"\ngcd(-24, 18) = {gcd(-24,18)}")
print(f"gcd(-12,-18) = {gcd(-12,-18)}")
print("""
观察：负数输入时，取模运算结果符号依赖Python规则，gcd返回正数；
数学上最大公约数定义针对非负整数，输入负数行为没有数学意义。
""")
```

```python
import math

#1 euler_limit 极限形式 (1+1/n)**n
def euler_limit(n):
    return (1 + 1.0 / n) ** n

#2 euler_infinite_sum 前n项级数和
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1
    for k in range(n):
        if k>0:
            fact *= k
        total += 1.0 / fact
    return total

#3 euler_approx：直到与真实e误差小于epsilon，while循环
def euler_approx(epsilon):
    real_e = math.exp(1)
    k = 0
    while True:
        approx = euler_infinite_sum(k)
        if abs(approx - real_e) < epsilon:
            return approx
        k += 1

#4 print_euler_sum_table，打印k=1~n每一行
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")


#测试
print("===euler_limit测试===")
print(f"euler_limit(1000) = {euler_limit(1000)}")

print("\n===euler_infinite_sum测试===")
print(f"n=0 → {euler_infinite_sum(0)}")
print(f"n=1 → {euler_infinite_sum(1)}")
print(f"n=4 → {euler_infinite_sum(4)}")

print("\n===euler_approx(1e-4) 误差小于0.0001===")
print(euler_approx(1e-4))
print(f"math.exp(1)真实e = {math.exp(1)}")

print("\n===print_euler_sum_table(6)===")
print_euler_sum_table(6)
```

```python
def is_power_of_2(n):
    """判断n是否为2的幂，只用%和//"""
    while n % 2 == 0:
        n = n // 2
    return n == 1

def is_power(b, n):
    """判断n是不是b的幂，b,n正整数"""
    if n == 1:
        return True # b**0 = 1
    while n % b == 0:
        n = n // b
    return n == 1

#测试
print("===is_power_of_2测试===")
test_2 = [1,2,4,8,10,32]
for num in test_2:
    print(f"{num} is power of 2 ? {is_power_of_2(num)}")

print("\n===is_power(b,n)测试===")
print(f"is_power(3,9) → {is_power(3,9)}")
print(f"is_power(5,125) → {is_power(5,125)}")
print(f"is_power(2,10) → {is_power(2,10)}")
print(f"is_power(7,1) → {is_power(7,1)}")
```
