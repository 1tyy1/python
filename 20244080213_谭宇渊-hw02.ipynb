```python
def is_palindrome(s: str) -> bool:
    # 忽略空格，全部转小写
    cleaned = s.replace(" ", "").lower()
    return cleaned == cleaned[::-1]

# 测试
print(is_palindrome("T A C O cat"))
print(is_palindrome("level"))
```

```python
def is_abecedarian(s: str) -> bool:
    cleaned = s.replace(" ", "").lower()
    for i in range(len(cleaned)-1):
        if cleaned[i] > cleaned[i+1]:
            return False
    return True

# 测试
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("beet"))
```

```python
def remove_vowels(s: str) -> str:
    vowels = {"a", "e", "i", "o", "u", "A", "E", "I", "O", "U"}
    res = []
    for ch in s:
        if ch not in vowels:
            res.append(ch)
    return "".join(res)

# 测试
print(remove_vowels('cat'))
print(remove_vowels('audio'))
```

```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1]

# 测试
print(list_reverse([1,2,3]))
```

```python
def is_sorted(seq):
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True

# 测试
print(is_sorted([1,2,3,4]))
print(is_sorted([1,3,2]))
```

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not all(isinstance(x, (int, float)) for x in t):
        raise TypeError("元组内元素必须为int或者float")
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是int或者float")
    return tuple(x * s for x in t)

# 测试
print(vec_scalar_mult((1,2,3), 2))
```

```python
def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("输入必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个向量元组长度必须相等")
    if not (all(isinstance(x, (int,float)) for x in t1) and all(isinstance(x,(int,float)) for x in t2)):
        raise TypeError("向量元素必须是int/float")
    total = 0.0
    for a,b in zip(t1,t2):
        total += a * b
    return total

# 测试
print(vec_inner_product((1,2),(3,4)))
```

```python
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return True
    row_len = len(mx[0])
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if len(row) != row_len:
            return False
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True

# 测试
print(check_valid_mx(((1,2),(3,4))))
```

```python
def mx_vec_mult(mat, vec):
    if not check_valid_mx(mat):
        raise ValueError("不是合法矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    if len(mat) == 0:
        return ()
    n_col = len(mat[0])
    if len(vec) != n_col:
        raise ValueError(f"矩阵列数{n_col}和向量维度{len(vec)}不匹配")
    result = []
    for row in mat:
        dot = vec_inner_product(row, vec)
        result.append(dot)
    return tuple(result)

# 测试
print(mx_vec_mult(((1,2),(3,4)), (10,20)))
```

```python
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int):
            raise TypeError("维度dim必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        self.dim = dim

        if elements is None:
            self.elements = [0.0 for _ in range(dim)]
        else:
            if not isinstance(elements, (list, tuple)):
                raise TypeError("elements必须是list或者tuple")
            self.elements = [float(x) for x in elements]
            if len(self.elements) != self.dim:
                raise ValueError(f"维度{dim}，但是元素数量{len(self.elements)}不匹配")

    # 3.比较运算符 __eq__ __lt__ __le__ __gt__ __ge__
    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("比较的两个向量维度不同")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("比较的两个向量维度不同")
        return self.elements < other.elements

    def __le__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("比较的两个向量维度不同")
        return self.elements <= other.elements

    def __gt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("比较的两个向量维度不同")
        return self.elements > other.elements

    def __ge__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("比较的两个向量维度不同")
        return self.elements >= other.elements

    #4 dot内积方法
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot参数必须是Vector实例")
        if self.dim != other.dim:
            raise ValueError("两个向量维度不一致，无法计算内积")
        s = 0.0
        for a,b in zip(self.elements, other.elements):
            s += a * b
        return s

    #5 __mul__ 右乘；__rmul__左乘，支持标量、向量逐元素相乘
    def __mul__(self, other):
        if isinstance(other, (int, float)):
            new_elems = [e * other for e in self.elements]
            return Vector(self.dim, new_elems)
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("向量相乘维度不一致")
            new_elems = [a*b for a,b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_elems)
        else:
            return NotImplemented

    def __rmul__(self, scalar):
        if isinstance(scalar, (int, float)):
            new_elems = [e * scalar for e in self.elements]
            return Vector(self.dim, new_elems)
        return NotImplemented

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

# 测试示例
v1 = Vector(2, (2,3))
v2 = Vector(2, (4,5))
print(v1 * v2)
print(2 * v1)
print(v1.dot(v2))
print(v1 < v2)
```
