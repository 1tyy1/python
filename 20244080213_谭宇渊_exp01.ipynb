```python
# 1 score_level
def score_level(score):
    if not (0 <= score <=100):
        raise ValueError("分数必须0‑100")
    if 90 <= score <=100:
        return "优秀"
    elif 80 <= score <90:
        return "良好"
    elif 60 <= score <80:
        return "及格"
    else:
        return "不及格"

print("score_level(95):", score_level(95))
print("score_level(72):", score_level(72))
# score_level(120) #取消注释测试抛异常

#2 for while求1‑100和
sum_for=0
for i in range(1,101):
    sum_for +=i
print(f"\nfor循环1‑100和：{sum_for}")

sum_while=0
n=1
while n<=100:
    sum_while +=n
    n +=1
print(f"while循环1‑100和：{sum_while}")

#3 阶乘：递归+迭代
def fact_rec(n):
    if n ==0:
        return 1
    return n*fact_rec(n-1)

def fact_iter(n):
    res=1
    for i in range(1,n+1):
        res *=i
    return res

print("\n阶乘对比(0,1,5,10)")
for x in [0,1,5,10]:
    print(f"n={x} 递归:{fact_rec(x)} 迭代:{fact_iter(x)}")

#4 summary 返回最小、最大、平均
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    mn = min(values)
    mx = max(values)
    avg = sum(values)/len(values)
    return mn, mx, avg

vals=[10,20,30,40]
mi,ma,av = summary(vals)
print(f"\nsummary解包：min={mi},max={ma},avg={av:.2f}")

#5 sum_all(*args)可变参数
def sum_all(*args):
    total=0
    for v in args:
        total +=v
    return total

print("\nsum_all测试")
print(f"无参 {sum_all()}")
print(f"单参 {sum_all(10)}")
print(f"多参 {sum_all(1,2,3,4,5)}")
```

```python
import os
import pickle

#1 工作目录，建文件夹
cwd = os.getcwd()
print(f"当前工作目录 {cwd}")
data_dir = os.path.join(cwd,"exp01_data")
os.makedirs(data_dir, exist_ok=True)

# 准备students.txt原始文本内容
content_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

stu_txt_path = os.path.join(data_dir,"students.txt")
#2写入文件
with open(stu_txt_path,"w",encoding="utf‑8") as f:
    f.write(content_text)

# parse_student 函数，后面要复制到student_utils.py
def parse_student(line):
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts)!=2:
        raise ValueError(f"字段数错误，得到{len(parts)}个字段")
    name,s_str = parts
    name = name.strip()
    if len(name)==0:
        raise ValueError("姓名为空")
    try:
        score = int(s_str.strip())
    except ValueError:
        raise ValueError(f"成绩不是整数:{s_str}")
    if not (0 <= score <= 100):
        raise ValueError(f"成绩越界0‑100：{score}")
    return {"name":name,"score":score}

#4逐行处理捕获异常
valid_records = []
error_log = []
with open(stu_txt_path,"r",encoding="utf‑8") as f:
    lines = f.readlines()
for lineno, raw in enumerate(lines, start=1):
    if lineno == 1:
        continue #跳过表头
    try:
        rec = parse_student(raw)
        valid_records.append(rec)
    except ValueError as err:
        error_log.append({"line":lineno,"err":str(err)})

print(f"\n有效记录：{valid_records}")
print(f"错误记录：{error_log}")

#统计
valid_num = len(valid_records)
invalid_num = len(error_log)
if valid_num>0:
    avg_score = sum(r["score"] for r in valid_records)/valid_num
    pass_count = sum(1 for r in valid_records if r["score"]>=60)
    pass_rate = pass_count / valid_num
else:
    avg_score=0
    pass_rate=0

#5 report.txt w模式覆盖，a追加
report_path = os.path.join(data_dir,"report.txt")
# w模式覆盖
with open(report_path,"w",encoding="utf‑8") as fw:
    fw.write(f"有效记录数:{valid_num}\n")
    fw.write(f"无效记录数:{invalid_num}\n")
    fw.write(f"平均分:{avg_score:.2f}\n")
    fw.write(f"及格率:{pass_rate:.1%}\n")

# a模式追加
with open(report_path,"a",encoding="utf‑8") as fa:
    fa.write("----使用a模式追加这一行----\n")

#6 pickle保存读取
pkl_path = os.path.join(data_dir,"students.pkl")
with open(pkl_path,"wb") as fwb:
    pickle.dump(valid_records, fwb)

with open(pkl_path,"rb") as frb:
    loaded_data = pickle.load(frb)
print("\npickle读取后和原数据相等？", loaded_data == valid_records)

#7 模块导入测试：需要把parse_student复制保存到同级 student_utils.py
print("\n===测试导入student_utils===")
try:
    from student_utils import parse_student
    test_ok = parse_student("Tom,77")
    print("有效测试：", test_ok)
    try:
        parse_student("Bad,200")
    except ValueError as e:
        print("无效测试捕获异常：",e)
except ModuleNotFoundError:
    print("提示：请把parse_student保存为 student_utils.py放在notebook同一文件夹！")

#8捕获FileNotFoundError
try:
    with open("not_exist_1234.txt","r",encoding="utf‑8") as f:
        pass
except FileNotFoundError as e:
    print(f"\n捕获文件不存在异常:{e}")
```

```python
import copy

#1 Student类
class Student:
    def __init__(self,name,score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >=60

    def __str__(self):
        return f"Student(name={self.name},score={self.score})"

#2 根据实验4有效记录创建Student对象
print("====Student实例====")
for r in valid_records:
    s_obj = Student(r["name"], r["score"])
    print(f"{s_obj}, 是否及格:{s_obj.is_passed()}")

#3 GraduateStudent继承
class GraduateStudent(Student):
    def __init__(self,name,score,research_topic):
        super().__init__(name,score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name},score={self.score},topic={self.research_topic})"

g_stu = GraduateStudent("LiMing",82,"大模型研究")
print("\n研究生子类：",g_stu)
print("研究生是否及格：", g_stu.is_passed())

#4 直接赋值 copy.copy copy.deepcopy
print("\n====深浅拷贝对比====")
origin = {"name":"Alice", "scores":[85,90]}
#直接赋值
ref = origin
#浅拷贝
shallow = copy.copy(origin)
#深拷贝
deep = copy.deepcopy(origin)

#修改内部列表
origin["scores"][0] = 999
print(f"原始origin:{origin}")
print(f"直接赋值ref:{ref}")
print(f"copy.copy浅拷贝shallow:{shallow}")
print(f"copy.deepcopy深拷贝deep:{deep}")
print("直接赋值：完全同一个对象；copy.copy只复制外层，内部列表共享；deepcopy递归全部复制")
```
